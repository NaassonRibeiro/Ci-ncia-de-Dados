```python
# Definindo Variáveis
fatias = [8,12,10]
atividades = ['Lazer','Estudo','Viagem']
colunas = ['r','m','y']

# Criando um gráfico
plt.pie(fatias, labels = atividades, colors= colunas, startangle = 90,
        shadow = True, explode = (0, 0.1, 0))

plt.show()
```

```python
# Definindo Variáveis
x = [1,3,5]
y = [1,2,5]

# Criando Gráfico
plt.plot(x,y)

# Atribuindo Título ao Gráfico
plt.title('Exemplo Utilizando Plot')
plt.xlabel('Variável 1')
plt.ylabel('Variável 2')

# Atribuindo Legenda
plt.plot(x,y,label='Uma Legenda')
plt.legend()

# Exibindo o Gráfico
plt.show()
```

```python
# Definindo Variáveis
x = [1,2,3,4,5,6,7,8]
y = [5,2,4,5,6,8,4,8]

# Criando Gráfico
plt.scatter(x, y, label='Pontos', color='b', marker= '*', s=100)
plt.legend()
plt.show()
```
